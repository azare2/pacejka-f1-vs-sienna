```python
import numpy as np
import matplotlib.pyplot as plt
import math
from enum import IntEnum
from typing import Tuple
import numba
from numba import njit
from numba.experimental import jitclass
```

```python
# Using ISO 8855 coordinate system where +x is forward,
# +y is left, and +z is up.

# Vehicle State Vector [x, y, v_x, v_y, theta, omega]
#
# x, y: Global position on the track in meters
#
# v_x, v_y: Velocity of car in its own local frame, m/s
# (v_x is forward speed, v_y is sideways slide speed)
#
# theta: Heading angle, radians. The direction the car's nose
# is pointing relative to the world.
#
# omega: Yaw rate, radians/second. How fast the car is
# spinning around its center. 
# omega > 0 means counter clockwise (left) turn.

# State vector index mapping
class StateIdx(IntEnum):
    X = 0
    Y = 1
    VX = 2
    VY = 3
    THETA = 4
    OMEGA = 5

def initialize_state(v_x_initial_kph: float) -> np.ndarray:
    """Initialize the 6-DoF state vector with starting v_x converted to m/s"""
    v_x_initial_mps = v_x_initial_kph / 3.6  # Convert kph to m/s
    
    # State: [x, y, v_x, v_y, theta, omega]
    state = np.array([
        0.0,              # x (starting at origin)
        0.0,              # y
        v_x_initial_mps,  # v_x (forward velocity)
        0.0,              # v_y (no initial sideways sliding)
        0.0,              # theta (pointing straight along x-axis)
        0.0               # omega (no initial spin/yaw rate)
    ], dtype=np.float64)

    return state
```

```python
# Numba vehicle jitclass and instances

vehicle_spec = [
    ('mass', numba.float64),          # curb weight + driver (kg)
    ('i_z', numba.float64),           # yaw moment of inertial (kg*m^2)
    ('wheelbase', numba.float64),     # distance between front and rear axles (m)
    ('track_width', numba.float64),   # distance between centers of left and right tires on the same axle (m)
    ('cg_to_front', numba.float64),   # distance from Center of Gravity (CoG) to front axle (m)
    ('cg_to_rear', numba.float64),    # distance from CoG to rear axle (m)
    ('cg_height', numba.float64),     # height of CoG above ground (m)
    ('cf', numba.float64),            # front tire cornering stiffness (N/rad), Pacejka simplified stiffness rate
    ('cr', numba.float64),            # rear tire cornering stiffness (N/rad)
    ('cl_area', numba.float64),         # Aerodynamic lift area/downforce area Cl*A (m^2)
]

@jitclass(vehicle_spec)
class NumbaVehicle:
    def __init__(self, mass, i_z, wheelbase, track_width, cg_to_front, cg_to_rear, cg_height, cf, cr, cl_area):
        self.mass = mass
        self.i_z = i_z
        self.wheelbase = wheelbase
        self.track_width = track_width
        self.cg_to_front = cg_to_front
        self.cg_to_rear = cg_to_rear
        self.cg_height = cg_height
        self.cf = cf
        self.cr = cr
        self.cl_area = cl_area
    
```

```python
# Estimated constants for a 2017 Toyota Sienna SE
sienna = NumbaVehicle(
    mass=2107.0,      # kg, curb weight of 4480lbs + 165lb driver
    i_z=3720.0,       # estimated yaw inertia for large minivan
    wheelbase=3.03,   # meters, ~119.3 inches
    track_width=1.72, # meters, ~67.7 in
    cg_to_front=1.45, # front-heavy bias, ~1.45m from front axle
    cg_to_rear=1.58,  # remaining distance to rear axle
    cg_height=0.66,   # 26.15 in, from NHTSA Rollover Stability Measurements Report 2022 
    cf=80000.0,
    cr=85000.0,
    cl_area=0.1       # positive: slight aerodynamic lift
)
```

```python
@njit
def calculate_slip_angles(state: np.ndarray, delta: float, vehicle: NumbaVehicle):
    """
    Returns front and rear slip angles in radians.
    delta: Steering angle at front wheels in radians.
    delta > 0: steering to the left (+y).
    slip angle alpha > 0: tire is pointed further left than its velocity vector.
    """
    v_x = state[StateIdx.VX]
    v_y = state[StateIdx.VY]
    omega = state[StateIdx.OMEGA]
    
    # Avoid division by zero if car is stopped
    if v_x < 0.1:
        return 0.0, 0.0
    
    # Using bicycle model with delta pointing in front wheel direction.
    # Front slip angle = steering angle minus velocity angle at front axle
    alpha_f = delta - math.atan2(v_y + vehicle.cg_to_front * omega, v_x)
    
    # Assume the rear wheel is aligned with car centerline 
    # (rear wheels don't steer), so delta at the rear wheel = 0.
    alpha_r = -math.atan2(v_y - vehicle.cg_to_rear * omega, v_x)
    
    return alpha_f, alpha_r
    
```

```python
@njit
def calculate_normal_forces(v_x: float, vehicle: NumbaVehicle):
    """
    Returns dynamic vertical load on front and rear axles (Newtons).
    Includes static weight distribution + speed-dependent 
    aerodynamic downforce/lift.
    """
    g = 9.81
    rho = 1.225  # air density at sea level (kg/m^3)
    
    # Static axle weight split
    f_z_front_static = vehicle.mass * g * (vehicle.cg_to_rear / vehicle.wheelbase)
    f_z_rear_static = vehicle.mass * g * (vehicle.cg_to_front / vehicle.wheelbase)
    
    # Dynamic aerodynamic force (F_aero = 0.5 * rho * ClA * v^2)
    # Split 50/50 between front and rear axles
    f_aero_axle = (0.5 * rho * vehicle.cl_area * (v_x**2)) / 2.0
    
    f_z_front = max(0.0, f_z_front_static - f_aero_axle)
    f_z_rear = max(0.0, f_z_rear_static - f_aero_axle)
    
    return f_z_front, f_z_rear
```

```python
@njit
def calculate_pacejka_lateral_force(alpha: float, f_z: float, c_alpha: float) -> float:
    """
    Returns lateral tire force F_y (Newtons).
    alpha: slip angle (rad)
    f_z: normal load (N)
    c_alpha: cornering stiffness parameter
    """
    if f_z <= 0.0:
        return 0.0
    
    # Pacejka coefficients
    # B: stiffness factor
    # C: shape factor
    # D: peak factor
    # E: curvature factor
    
    # Standard Magic Formula shape parameters for dry asphalt
    C = 1.30      # for lateral force
    E = -1.0
    mu = 1.1      # peak friction coefficient (mu ~ 1.0 - 1.6)
    
    D = mu * f_z  # peak available force
    
    # Avoid division by zero
    if D <= 0.0:
        return 0.0
    
    B = c_alpha / (C * D)
    
    # Pacejka Magic Formula
    f_y = D * math.sin(C * math.atan(B * alpha - E * (B * alpha - math.atan(B * alpha))))
    
    return f_y
    
```

```python
@njit
def vehicle_dynamics(state: np.ndarray, delta:float, vehicle: NumbaVehicle) -> np.ndarray:
    """
    Calculates time derivatives of state vector, d(state)/dt
    Returns: [dx/dt, dy/dt, dv_x/dt, dv_y/dt, dtheta/dt, domega/dt]
    State: [x, y, v_x, v_y, theta, omega]
    delta: Steering angle at front wheels in radians.
    """
    v_x = state[StateIdx.VX]
    v_y = state[StateIdx.VY]
    theta = state[StateIdx.THETA]
    omega = state[StateIdx.OMEGA]
    
    # Slip angles and forces
    alpha_f, alpha_r = calculate_slip_angles(state, delta, vehicle)
    f_z_f, f_z_r = calculate_normal_forces(v_x, vehicle)
    
    f_y_f = calculate_pacejka_lateral_force(alpha_f, f_z_f, vehicle.cf)
    f_y_r = calculate_pacejka_lateral_force(alpha_r, f_z_r, vehicle.cr)
    
    # Assume zero throttle/braking drive force for pure cornering tests
    f_x_f = 0.0
    f_x_r = 0.0
    
    # Accelerations in the Car Body Frame
    
    # From Newton's 2nd Law and Transport Theorem:
    # Total longitudinal acceleration
    dv_x = ((f_x_r + f_x_f * math.cos(delta) - f_y_f * math.sin(delta)) / vehicle.mass) + (v_y * omega)
    
    # Total lateral acceleration
    dv_y = ((f_y_r + f_y_f * math.cos(delta) + f_x_f * math.sin(delta)) / vehicle.mass) - (v_x * omega)
    
    # Yaw angular acceleration. 
    # From 2D scalar Euler Equation for rotational motion.
    domega = (vehicle.cg_to_front * (f_y_f * math.cos(delta) + f_x_f * math.sin(delta)) - vehicle.cg_to_rear * f_y_r) / vehicle.i_z
    
    # Global Frame (e.g. racetrack coordinate system) velocities:
    # Recall that theta is the heading angle (rad), the direction the
    # car is pointing relative to the world. v_x, v_y are velocity
    # components in the car's local rotating body frame. We apply the
    # 2D rotation matrix (rotate by theta) to get the car's velocity
    # in the Global racetrack axes.
    dx = v_x * math.cos(theta) - v_y * math.sin(theta)
    dy = v_x * math.sin(theta) + v_y * math.cos(theta)
    dtheta = omega
    
    return np.array([dx, dy, dv_x, dv_y, dtheta, domega], dtype=np.float64)
```
